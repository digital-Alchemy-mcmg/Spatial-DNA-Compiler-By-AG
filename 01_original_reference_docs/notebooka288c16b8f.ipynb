```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from core_engine import process_resume, format_report_text

text_area = widgets.Textarea(
    value='',
    placeholder='Paste your entire document or resume here...',
    description='Document:',
    layout=widgets.Layout(width='100%', height='300px')
)

process_button = widgets.Button(
    description='Process Text',
    button_style='success',
    icon='check'
)

output_area = widgets.Output()

def on_button_clicked(b):
    with output_area:
        clear_output()
        raw_text = text_area.value
        
        if not raw_text.strip():
            print("The text box is empty. Please paste your document first.")
            return
        
        print("Scanning and processing deterministically...
")
        results = process_resume(raw_text) 
        report = format_report_text(results)
        print(report)

process_button.on_click(on_button_clicked)

display(widgets.VBox([
    widgets.HTML("<b>Step 1: Paste your text below. Step 2: Click 'Process Text'</b>"),
    text_area,
    process_button,
    output_area
]))
```
